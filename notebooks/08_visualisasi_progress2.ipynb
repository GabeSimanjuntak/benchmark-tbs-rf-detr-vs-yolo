```python
# Membaca distribusi kelas berdasarkan split
pabrik_split = pd.read_csv(
    RESULTS_DIR / "eda_dataset_pabrik" /
    "distribusi_kelas_per_split.csv"
)

kalteng_split = pd.read_csv(
    RESULTS_DIR / "eda_dataset_perkebunan" /
    "distribusi_kelas_per_split.csv"
)


def plot_split_distribution(
    dataframe,
    dataset_name,
    class_column,
    output_filename
):
    pivot_df = dataframe.pivot(
        index=class_column,
        columns="Split",
        values="Jumlah Objek"
    ).fillna(0)

    # Pastikan urutan split konsisten
    split_order = [
        split for split in ["train", "valid", "test"]
        if split in pivot_df.columns
    ]

    pivot_df = pivot_df[split_order]

    ax = pivot_df.plot(
        kind="bar",
        figsize=(12, 6),
        width=0.8
    )

    ax.set_title(
        f"Distribusi Kelas per Split - {dataset_name}"
    )
    ax.set_xlabel("Kelas")
    ax.set_ylabel("Jumlah Objek")
    ax.set_xticklabels(
        ax.get_xticklabels(),
        rotation=30,
        ha="right"
    )
    ax.legend(title="Split")
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / output_filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# Pabrik Kalsel
plot_split_distribution(
    pabrik_split,
    "Pabrik Kalsel",
    "Nama Kelas",
    "distribusi_split_pabrik.png"
)

# Perkebunan Kalteng
plot_split_distribution(
    kalteng_split,
    "Perkebunan Kalteng",
    "Nama Kelas",
    "distribusi_split_kalteng.png"
)
```

```python
from pathlib import Path
from collections import Counter
import pandas as pd

FILTERED_DATASETS = {
    "Pabrik Kalsel": (
        PROJECT_DIR / "datasets" / "processed" / "filtered"
        / "dataset-pabrik-kalsel"
    ),
    "Perkebunan Kalteng": (
        PROJECT_DIR / "datasets" / "processed" / "filtered"
        / "dataset-perkebunan-kalteng"
    ),
    "Malaysia v7": (
        PROJECT_DIR / "datasets" / "processed"
        / "dataset-malaysia-v7-filtered"
    ),
}

CLASS_NAMES = {
    0: "Unripe",
    1: "Ripe"
}

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

summary = []

for dataset_name, dataset_dir in FILTERED_DATASETS.items():

    print(f"\nMemeriksa: {dataset_name}")

    for split in ["train", "valid", "test"]:

        image_dir = dataset_dir / split / "images"
        label_dir = dataset_dir / split / "labels"

        if not image_dir.exists() or not label_dir.exists():
            print(f"  {split}: tidak tersedia")
            continue

        images = [
            p for p in image_dir.iterdir()
            if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        ]

        class_counts = Counter()
        missing_labels = []
        total_boxes = 0

        for image_path in images:

            label_path = label_dir / f"{image_path.stem}.txt"

            if not label_path.exists():
                missing_labels.append(image_path.name)
                continue

            for line in label_path.read_text(
                encoding="utf-8"
            ).splitlines():

                if not line.strip():
                    continue

                class_id = int(float(line.split()[0]))

                if class_id not in CLASS_NAMES:
                    raise ValueError(
                        f"Class ID tidak valid: {class_id} "
                        f"pada {label_path}"
                    )

                class_counts[class_id] += 1
                total_boxes += 1

        summary.append({
            "Dataset": dataset_name,
            "Split": split,
            "Jumlah Gambar": len(images),
            "Unripe": class_counts[0],
            "Ripe": class_counts[1],
            "Total Bounding Box": total_boxes,
            "Label Tidak Ditemukan": len(missing_labels)
        })

df_harmonized = pd.DataFrame(summary)

display(df_harmonized)

output_path = (
    RESULTS_DIR / "visualisasi_progress2"
    / "distribusi_harmonisasi_lodo_v2.csv"
)

df_harmonized.to_csv(output_path, index=False)

print("\nStatistik disimpan di:")
print(output_path)
```

```python
# Rekapitulasi total objek per dataset
harmonized_total = (
    df_harmonized
    .groupby("Dataset")[["Unripe", "Ripe"]]
    .sum()
)

dataset_order = [
    "Pabrik Kalsel",
    "Perkebunan Kalteng",
    "Malaysia v7"
]

harmonized_total = harmonized_total.reindex(dataset_order)

# Grafik jumlah bounding box
ax = harmonized_total.plot(
    kind="bar",
    figsize=(10, 6),
    width=0.75
)

ax.set_title("Distribusi Kelas Harmonisasi pada Tiga Dataset")
ax.set_xlabel("Dataset")
ax.set_ylabel("Jumlah Bounding Box")
ax.set_xticklabels(
    ["Pabrik Kalsel", "Perkebunan Kalteng", "Malaysia v7"],
    rotation=15,
    ha="right"
)
ax.legend(title="Kelas")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "distribusi_kelas_harmonisasi.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# Grafik persentase kelas per dataset
harmonized_percentage = (
    harmonized_total
    .div(harmonized_total.sum(axis=1), axis=0)
    * 100
)

ax = harmonized_percentage.plot(
    kind="bar",
    figsize=(10, 6),
    width=0.75
)

ax.set_title("Persentase Kelas Harmonisasi pada Tiga Dataset")
ax.set_xlabel("Dataset")
ax.set_ylabel("Persentase Bounding Box (%)")
ax.set_xticklabels(
    ["Pabrik Kalsel", "Perkebunan Kalteng", "Malaysia v7"],
    rotation=15,
    ha="right"
)
ax.legend(title="Kelas")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "persentase_kelas_harmonisasi.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import random
import cv2
import matplotlib.pyplot as plt
from pathlib import Path

# Konfigurasi dataset hasil filtering
DATASETS = {
    "Pabrik Kalsel": (
        PROJECT_DIR / "datasets" / "processed" / "filtered"
        / "dataset-pabrik-kalsel"
    ),
    "Perkebunan Kalteng": (
        PROJECT_DIR / "datasets" / "processed" / "filtered"
        / "dataset-perkebunan-kalteng"
    ),
    "Malaysia v7": (
        PROJECT_DIR / "datasets" / "processed"
        / "dataset-malaysia-v7-filtered"
    ),
}

CLASS_NAMES = {
    0: "Unripe",
    1: "Ripe"
}

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

random.seed(42)

# Ambil sampel dari split train setiap dataset
samples = []

for dataset_name, dataset_dir in DATASETS.items():

    image_dir = dataset_dir / "train" / "images"
    label_dir = dataset_dir / "train" / "labels"

    if not image_dir.exists() or not label_dir.exists():
        raise FileNotFoundError(
            f"Folder images/labels tidak ditemukan: {dataset_name}"
        )

    images = [
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    for class_id, class_name in CLASS_NAMES.items():

        candidates = []

        for image_path in images:
            label_path = label_dir / f"{image_path.stem}.txt"

            if not label_path.exists():
                continue

            lines = [
                line.strip()
                for line in label_path.read_text(
                    encoding="utf-8"
                ).splitlines()
                if line.strip()
            ]

            if any(
                int(float(line.split()[0])) == class_id
                for line in lines
            ):
                candidates.append(image_path)

        if not candidates:
            raise ValueError(
                f"Tidak ditemukan sampel {class_name} "
                f"pada {dataset_name}"
            )

        selected_image = random.choice(candidates)

        samples.append({
            "dataset": dataset_name,
            "class_id": class_id,
            "class_name": class_name,
            "image_path": selected_image,
            "label_path": (
                label_dir / f"{selected_image.stem}.txt"
            )
        })


# Visualisasi enam sampel
fig, axes = plt.subplots(2, 3, figsize=(18, 11))
axes = axes.flatten()

for ax, sample in zip(axes, samples):

    image = cv2.imread(str(sample["image_path"]))

    if image is None:
        raise ValueError(
            f"Gambar gagal dibaca: {sample['image_path']}"
        )

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    height, width = image.shape[:2]

    lines = sample["label_path"].read_text(
        encoding="utf-8"
    ).splitlines()

    for line in lines:

        if not line.strip():
            continue

        values = line.split()
        class_id = int(float(values[0]))

        xc, yc, bw, bh = map(float, values[1:5])

        x1 = int((xc - bw / 2) * width)
        y1 = int((yc - bh / 2) * height)
        x2 = int((xc + bw / 2) * width)
        y2 = int((yc + bh / 2) * height)

        cv2.rectangle(
            image,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

        label = CLASS_NAMES[class_id]

        cv2.putText(
            image,
            label,
            (x1, max(y1 - 8, 15)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

    ax.imshow(image)
    ax.set_title(
        f"{sample['dataset']} - {sample['class_name']}",
        fontsize=11
    )
    ax.axis("off")

plt.suptitle(
    "Visual Comparison of Harmonized Classes Across Domains",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "visualisasi_harmonisasi_lintas_domain.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Visualisasi berhasil disimpan.")
print(
    OUTPUT_DIR / "visualisasi_harmonisasi_lintas_domain.png"
)

for sample in samples:
    print(
        sample["dataset"],
        "-",
        sample["class_name"],
        "-",
        sample["image_path"].name
    )
```
